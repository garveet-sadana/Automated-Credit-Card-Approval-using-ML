IMPORTING LIBRARIES

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
from sklearn.model_selection import cross_val_score
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
```

DEFINING DATASET

```python
df = pd.read_csv('/content/application_record.csv')
df1 = pd.read_csv('/content/credit_record.csv')
```

READING & ANALYSING DATASET

```python
nature = []
for i in df1['STATUS']:
  if i == "X":
    nature.append(0)
  elif i == "C" or i == "0" or i == "1" or i == "2":
    nature.append(1)
  else:
    nature.append(0)
```

```python
df1['COUSTOMER_NATURE'] = nature
```

MERGING DATASET

```python
final_data = df.merge(df1)
```

EXPLORATORY DATA ANALYISIS (CLEANING DATASET)

```python
final_data = final_data.drop("NAME_FAMILY_STATUS",axis= 1)
final_data = final_data.drop("FLAG_MOBIL",axis= 1)
final_data = final_data.drop("FLAG_WORK_PHONE",axis= 1)
final_data = final_data.drop("FLAG_PHONE",axis= 1)
final_data = final_data.drop("FLAG_EMAIL",axis= 1)
final_data = final_data.drop("MONTHS_BALANCE",axis= 1)
final_data = final_data.drop("STATUS",axis= 1)
```

```python
final_data.info()
```

FEATURE ENGINEERING (ENCODING FEATURES)

```python
final_data["CODE_GENDER"] = final_data["CODE_GENDER"].map({"F":0, "M":1})
```

```python
final_data["FLAG_OWN_CAR"] = pd.get_dummies(final_data["FLAG_OWN_CAR"],drop_first = 1)
```

```python
final_data["FLAG_OWN_REALTY"] = pd.get_dummies(final_data["FLAG_OWN_REALTY"],drop_first = 1)
```

```python
final_data["NAME_INCOME_TYPE"] = final_data["NAME_INCOME_TYPE"].map({"Working":0, "Commercial associate":1, "Pensioner":2,"State servant":3,"Student":4})
```

```python
final_data["NAME_EDUCATION_TYPE"] = final_data["NAME_EDUCATION_TYPE"].map({"Higher education":0, "Secondary / secondary special":1, "Incomplete higher":2,"Lower secondary":3,"Academic degree":4})
```

```python
final_data["NAME_HOUSING_TYPE"] = final_data["NAME_HOUSING_TYPE"].map({"Rented apartment":0, "House / apartment":1, "Municipal apartment":2,"With parents":3,"Co-op apartment":4,"Office apartment":5})
```

```python
def impute_nan(final_data,variable):
    final_data[variable+"NEW"]=np.where(final_data[variable].isnull(),"Missing",final_data[variable])
```

```python
for feature in ['OCCUPATION_TYPE']:
    impute_nan(final_data,feature)
```

```python
final_data["OCCUPATION_TYPENEW"].unique()
```

```python
final_data["OCCUPATION_TYPENEW"] = final_data["OCCUPATION_TYPENEW"].map({"Missing":0, "Security staff":1, "Sales staff":2,"Accountants":3,"Laborers":4,"Managers":5,"Drivers":6,"Core staff":7,"High skill tech staff":8,"Cleaning staff":9,"Private service staff":10,"Cooking staff":11,"Low-skill Laborers":12,"Medicine staff":13,"Secretaries":14,"Waiters/barmen staff":15,"HR staff":16,"Realty agents":17,"IT staff":18})
```

```python
final_data["OCCUPATION_TYPENEW"].unique()
```

```python
final_data = final_data.drop("OCCUPATION_TYPE",axis= 1)
```

DROPPING DUPLICATE DATA

```python
final_data=final_data.drop_duplicates(subset = "ID")
```

```python
final_data.info()
```

```python
final_data["COUSTOMER_NATURE1"]=final_data["COUSTOMER_NATURE"]

```

```python
final_data = final_data.drop("COUSTOMER_NATURE",axis= 1)
```
